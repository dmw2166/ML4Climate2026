# Week 4 live coding: classification and regression

A 45-minute tutorial to type through together. It covers the four supervised methods from
this week's lecture, linear regression, logistic regression, softmax regression, and support
vector machines, on one small dataset, and it introduces the habit that matters more than any
of them: always compare against a baseline.

We work with **daily weather at Central Park, New York** from NOAA, 1990 to the present. The
question is the oldest one in forecasting: given today's weather, what happens tomorrow?
Assignment 4 asks the same methods of a different dataset, tropical cyclone records, so
everything here transfers and nothing here is the answer to the assignment.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn

print("pandas ", pd.__version__)
print("sklearn", sklearn.__version__)
```

## 1. Load the data and frame the problem

NOAA's daily summaries are available through a URL, no account needed. Station `USW00094728`
is Central Park. We ask for maximum and minimum temperature (°C), precipitation (mm), and snowfall (mm).

```python
url = ("https://www.ncei.noaa.gov/access/services/data/v1"
       "?dataset=daily-summaries&stations=USW00094728"
       "&startDate=1990-01-01&endDate=2025-12-31"
       "&dataTypes=TMAX,TMIN,PRCP,SNOW&format=csv&units=metric")
wx = pd.read_csv(url, parse_dates=["DATE"])

print(wx.shape)
print("from", wx["DATE"].min().date(), "to", wx["DATE"].max().date())
wx.head()
```

Look before you model. Ranges tell you the units are what you think, and the NaN count tells
you whether anything is missing.

```python
print(wx[["TMAX", "TMIN", "PRCP"]].describe().round(1))
print()
print("missing values per column:")
print(wx.isna().sum())
```

**Supervised learning needs a label for every row.** Our labels are about *tomorrow*, so we
shift each column back by one day: the value in row $t$ of `TMAX_tomorrow` is the value of
`TMAX` in row $t+1$. The last row has no tomorrow and gets dropped.

Notice that the same dataset gives us both kinds of label from the lecture:

| Label | Type | Task |
| --- | --- | --- |
| `TMAX_tomorrow` | a number, °C | regression |
| `rain_tomorrow` | yes or no | classification |

```python
wx = wx.sort_values("DATE").reset_index(drop=True)

wx["TMAX_tomorrow"] = wx["TMAX"].shift(-1)
wx["PRCP_tomorrow"] = wx["PRCP"].shift(-1)
wx["rain_tomorrow"] = (wx["PRCP_tomorrow"] > 0).astype(int)   # 1 = rain, 0 = dry

wx = wx.dropna().reset_index(drop=True)

print("rows:", len(wx))
print("fraction of days followed by rain:", wx["rain_tomorrow"].mean().round(3))
wx[["DATE", "TMAX", "TMIN", "PRCP", "TMAX_tomorrow", "rain_tomorrow"]].head()
```

**Features.** Today's high, low, and rainfall are the obvious inputs. Season matters too, but
the day of the year is a poor feature as a raw number: December 31 and January 1 are 364 apart
on that scale and one day apart in reality. Encoding it as a sine and cosine pair puts it on a
circle, which is the standard trick for any cyclic variable (hour of day, wind direction,
month).

```python
doy = wx["DATE"].dt.dayofyear
wx["doy_sin"] = np.sin(2 * np.pi * doy / 365.25)
wx["doy_cos"] = np.cos(2 * np.pi * doy / 365.25)

features = ["TMAX", "TMIN", "PRCP", "doy_sin", "doy_cos"]
wx[features].head()
```

## 2. Split by time, not at random

`train_test_split` shuffles rows. For a time series that leaks the answer: the model would be
tested on a Tuesday after training on the Monday and Wednesday around it, and consecutive days
are strongly correlated. We hold out the **last eight years** instead. The model never sees
any day after 2017, which is the situation a real forecast is in.

```python
train = wx[wx["DATE"] < "2018-01-01"]
test = wx[wx["DATE"] >= "2018-01-01"]

X_train, X_test = train[features].values, test[features].values
print("train:", len(train), "days,", train["DATE"].min().year, "to", train["DATE"].max().year)
print("test: ", len(test), "days,", test["DATE"].min().year, "to", test["DATE"].max().year)
```

## 3. Linear regression: tomorrow's high temperature

`LinearRegression` fits $\hat{y} = w_0 + w_1 x_1 + \dots + w_5 x_5$ by least squares. Three
lines: create the model, `fit` on training data, `predict` on test data. Every model in
scikit-learn has this same shape, which is why swapping methods later is so cheap.

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

y_train, y_test = train["TMAX_tomorrow"].values, test["TMAX_tomorrow"].values

reg = LinearRegression()
reg.fit(X_train, y_train)
pred = reg.predict(X_test)

rmse = np.sqrt(mean_squared_error(y_test, pred))
print(f"RMSE on the test years: {rmse:.2f} °C")
print(f"R^2 on the test years:  {r2_score(y_test, pred):.3f}")
```

An RMSE under 4 °C and an $R^2$ of 0.86 sound respectable. **They mean nothing on their own.** Skill is only ever
relative to something simpler. Two baselines every forecast has to beat:

- **Persistence**: tomorrow will be like today.
- **Climatology**: tomorrow will be the average for that calendar day, computed from the
  training years only.

```python
# Persistence: predict today's TMAX for tomorrow.
persist = test["TMAX"].values

# Climatology: mean TMAX_tomorrow for each day of year, learned on the training years.
clim_table = train.groupby(train["DATE"].dt.dayofyear)["TMAX_tomorrow"].mean()
clim = clim_table.reindex(test["DATE"].dt.dayofyear).values

for name, p in [("persistence", persist), ("climatology", clim), ("linear regression", pred)]:
    print(f"{name:18s} RMSE = {np.sqrt(mean_squared_error(y_test, p)):.2f} °C")
```

The regression beats both, by about 0.6 °C over persistence. That margin is the model's
actual contribution. Had it come out at 4.3 °C we would have learned nothing, no matter how
good the number looked in isolation.

**Reading the coefficients.** With features on different scales (°C, mm, and unitless sines)
the raw weights are not comparable. Standardize first, then the weight says how much the
prediction moves per standard deviation of that feature.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler().fit(X_train)
reg_std = LinearRegression().fit(scaler.transform(X_train), y_train)

for name, w in zip(features, reg_std.coef_):
    print(f"{name:8s} {w:+6.2f} °C per standard deviation")
```

Today's high dominates, which is persistence learned from data, with today's low adding
about half as much. The cosine term carries the seasonal cycle. Rain today pushes tomorrow
slightly cooler.

**Always plot predicted against observed.** The 1:1 line is where a perfect model sits.

```python
fig, ax = plt.subplots(figsize=(5.5, 5.5))
ax.scatter(y_test, pred, s=4, alpha=0.3)
lims = [-15, 42]
ax.plot(lims, lims, "k--", lw=1, label="1:1")
ax.set_xlim(lims); ax.set_ylim(lims)
ax.set_xlabel("Observed TMAX tomorrow (°C)")
ax.set_ylabel("Predicted TMAX tomorrow (°C)")
ax.set_title("Linear regression, test years 2018 onward")
ax.legend()
ax.grid(alpha=0.3)
plt.show()
```

The cloud is tilted slightly off the 1:1 line: hot days are under-predicted and cold days
over-predicted. Least squares pulls toward the mean, and one day of history is not enough to
know whether a heat wave will hold. Keep that in mind for Assignment 4, where the extremes are
exactly the part of the range people care about.

## 4. Logistic regression: rain or snow?

Same three lines, different model, different label. `LogisticRegression` fits a straight line
too, but passes it through a sigmoid so the output is a **probability** between 0 and 1.
Calling `predict` thresholds that probability at 0.5. Calling `predict_proba` gives you the
number itself, which is usually what you want.

The obvious label is "will it rain tomorrow?". Try it first, because it teaches the most
important habit in classification. Two new things in the cell: `make_pipeline` bundles the
scaler and the model into one object, so `fit` learns both from the training rows and `predict`
applies both, and the scaling can never be forgotten or fitted on test data by accident.


```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.metrics import accuracy_score, confusion_matrix, precision_score, recall_score

c_train, c_test = train["rain_tomorrow"].values, test["rain_tomorrow"].values

rain_clf = make_pipeline(StandardScaler(), LogisticRegression())
rain_clf.fit(X_train, c_train)
rain_label = rain_clf.predict(X_test)

always_dry = np.zeros_like(c_test)
print(f"always dry:          accuracy = {accuracy_score(c_test, always_dry):.3f}")
print(f"logistic regression: accuracy = {accuracy_score(c_test, rain_label):.3f}")
print()
print("confusion matrix (rows = actual dry, rain; columns = predicted dry, rain):")
print(confusion_matrix(c_test, rain_label))
print()
print(f"recall (of the rainy days, how many did it catch): {recall_score(c_test, rain_label):.3f}")
```

**The accuracy trap.** About two thirds of days are dry, so "dry every day" is right two
thirds of the time, and the model beats that by half a percentage point. The confusion matrix
shows why: it almost never says rain and catches fewer than one rainy day in ten. Today's
weather at one station says very little about tomorrow's rain. That is a fact about the
atmosphere, not a bug, and a real forecast uses the upstream weather map, which we have not
given it.

So ask a question the features can answer. On a day with precipitation, **is it rain or snow?**
Temperature decides that, and temperature is what we have. It is a diagnostic question rather
than a forecast, but the recipe is identical: build a label, split by time, `fit`, `predict`,
score against a baseline. We keep only the days with precipitation and label the ones where
snow fell.


```python
wet = wx[(wx["PRCP"] > 0) | (wx["SNOW"] > 0)].copy()
wet["snow"] = (wet["SNOW"] > 0).astype(int)        # 1 = snow fell, 0 = rain only

wet_train = wet[wet["DATE"] < "2018-01-01"]
wet_test = wet[wet["DATE"] >= "2018-01-01"]
Xw_train, Xw_test = wet_train[features].values, wet_test[features].values
s_train, s_test = wet_train["snow"].values, wet_test["snow"].values

print("precipitation days: train", len(wet_train), " test", len(wet_test))
print("fraction with snow in the test years:", s_test.mean().round(3))
```

```python
snow_clf = make_pipeline(StandardScaler(), LogisticRegression())
snow_clf.fit(Xw_train, s_train)

prob = snow_clf.predict_proba(Xw_test)[:, 1]        # column 1 = P(snow)
label = (prob >= 0.5).astype(int)

always_rain = np.zeros_like(s_test)
print(f"always rain:         accuracy = {accuracy_score(s_test, always_rain):.3f}")
print(f"logistic regression: accuracy = {accuracy_score(s_test, label):.3f}")
print()
print("confusion matrix (rows = actual rain, snow; columns = predicted rain, snow):")
print(confusion_matrix(s_test, label))
print()
print(f"precision (when it says snow, how often is it right): {precision_score(s_test, label):.3f}")
print(f"recall    (of the snow days, how many did it catch):  {recall_score(s_test, label):.3f}")
```

Now there is skill to talk about: 96.5% against a 93% baseline, and when the model says snow
it is right nearly nine times in ten. It still misses four snow days in ten, because snow is
7% of the precipitation days and the default threshold favors the common class. Same trap as
before, smaller.

**Look at the probabilities, not just the labels.** Plotting $P(\text{snow})$ against the
day's low temperature shows the sigmoid doing its job.


```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.scatter(wet_test["TMIN"][s_test == 0], prob[s_test == 0], s=8, alpha=0.4, label="rain days")
ax.scatter(wet_test["TMIN"][s_test == 1], prob[s_test == 1], s=12, alpha=0.7, color="tab:red", label="snow days")
ax.axhline(0.5, color="k", ls="--", lw=1, label="0.5 threshold")
ax.set_xlim(-20, 30)
ax.set_xlabel("Low temperature that day (°C)")
ax.set_ylabel("Predicted P(snow)")
ax.set_title("Logistic regression output on the test years")
ax.legend()
ax.grid(alpha=0.3)
plt.show()
```

A clean S-curve. The model is sure above about 5 °C and below about −5 °C, and all the
uncertainty lives in the few degrees around freezing, which is exactly where forecasters earn
their pay. The misses are the red points below the dashed line: snow on days whose low was
above freezing, which happens when the snow falls overnight into a day that warms.

The 0.5 threshold is a choice, not a law. A road crew deciding whether to salt wants to catch
every snow day and can tolerate false alarms. You can lower the threshold, or ask the model to
weight the two classes equally during training:


```python
snow_bal = make_pipeline(StandardScaler(), LogisticRegression(class_weight="balanced"))
snow_bal.fit(Xw_train, s_train)
label_bal = snow_bal.predict(Xw_test)

print(f"balanced weights: accuracy = {accuracy_score(s_test, label_bal):.3f}, "
      f"precision = {precision_score(s_test, label_bal):.3f}, recall = {recall_score(s_test, label_bal):.3f}")
print(confusion_matrix(s_test, label_bal))
```

Recall jumps from 0.58 to 0.93 and precision falls from 0.88 to 0.58: the model now catches
almost every snow day and raises roughly one false alarm for each real one. Accuracy went
down, and for the road crew the forecast got better. Which version you want is a decision
about the application, not about the algorithm.


## 5. Softmax regression: more than two classes

Rain versus snow throws away how much snow. Split the precipitation days into three classes:
rain, light snow (under 5 cm), and heavy snow (5 cm or more). `LogisticRegression` handles
this automatically: with more than two classes it fits one linear score per class and passes
them through a softmax, so the probabilities sum to one.


```python
def snow_class(mm):
    return np.where(mm == 0, 0, np.where(mm < 50, 1, 2))     # SNOW is in mm: 50 mm = 5 cm

m_train = snow_class(wet_train["SNOW"].values)
m_test = snow_class(wet_test["SNOW"].values)
names = ["rain", "light snow", "heavy snow"]

print("class balance in training years:", np.bincount(m_train) / len(m_train))

soft = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
soft.fit(Xw_train, m_train)
m_pred = soft.predict(Xw_test)

print()
print(f"accuracy: {accuracy_score(m_test, m_pred):.3f}   (always 'rain' would score {np.mean(m_test == 0):.3f})")
print()
print("confusion matrix, rows = actual, columns = predicted, order rain / light / heavy:")
print(confusion_matrix(m_test, m_pred))
```

Read the last row. Of the heavy snow days in eight test years, the model called fewer than a
third, and it mislabels most light snow as rain too. Heavy snow is 4% of the data, and a model
rewarded for overall accuracy learns to say "rain". This is **class imbalance**, and it is the
normal state of affairs in environmental data: floods, heat waves, and category 5 storms are
rare by definition. There is a second problem here as well: temperature separates snow from
rain, but it cannot say how much will fall. That depends on the storm, which these features do
not see.

One blunt fix for the imbalance is to weight the classes so that each contributes equally to
the loss.


```python
soft_bal = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000, class_weight="balanced"))
soft_bal.fit(Xw_train, m_train)
m_pred_bal = soft_bal.predict(Xw_test)

print(f"accuracy with balanced weights: {accuracy_score(m_test, m_pred_bal):.3f}")
print(confusion_matrix(m_test, m_pred_bal))
```

The light and heavy rows fill in, heavy snow recall more than doubles, and the price is about
sixty rain days flagged as snow. Accuracy fell back to the baseline. Neither version is
"right". Which you want depends on the cost of a miss versus the cost of a false alarm, and
that is a decision about the application, not about the algorithm. Week 6 gives you the tools
to make it properly.


## 6. Support vector machines

A linear model can only draw a straight boundary through feature space. A support vector
machine with an RBF kernel can draw a curved one. The scikit-learn call is the same shape as
before, which is the point of the library. We fit both kernels on the rain-or-snow task and put
logistic regression beside them.


```python
from sklearn.svm import SVC

for kernel in ["linear", "rbf"]:
    svm = make_pipeline(StandardScaler(), SVC(kernel=kernel, C=1.0))
    svm.fit(Xw_train, s_train)
    svm_label = svm.predict(Xw_test)
    print(f"{kernel:6s} SVM:          accuracy = {accuracy_score(s_test, svm_label):.3f}, "
          f"precision = {precision_score(s_test, svm_label):.3f}, recall = {recall_score(s_test, svm_label):.3f}")

print(f"logistic regression: accuracy = {accuracy_score(s_test, label):.3f}, "
      f"precision = {precision_score(s_test, label):.3f}, recall = {recall_score(s_test, label):.3f}")
```

Barely any different, and the curved boundary buys almost nothing over the straight one. The
border between rain and snow is close to a line in temperature, so there was nothing for the
extra flexibility to find. That is a general lesson: reach for a more complex model when a
simple one shows a specific failure you can name, not by default.

Two practical notes on SVMs. Training time grows roughly with the square of the number of
rows, so on a million-row dataset this cell would not finish. And a plain `SVC` gives you a
class, not a probability. If you need $P(\text{snow})$ for a decision, logistic regression
already has it.
